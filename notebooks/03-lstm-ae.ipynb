# 03 · LSTM autoencoder — deferred

The ROADMAP names the LSTM autoencoder **the first thing to cut if Phase 3 slips**, and it is deferred here on purpose:

- The rule-vs-IsolationForest table (notebook 04, ADR-0004) already delivers Phase 3's core argument: does a model beat an honest rule, and where.
- torch is a heavy workspace dependency; adding it before the comparison demands it would be scope for its own sake.

**When it lands:** a CPU-sized autoencoder trained on normal windows (same `X_train` split as notebook 02), reconstruction error as the score, `choose_threshold` on held-out normal, saved as `lstm-ae-v1` with a manifest — a drop-in third column for the evaluation table. The `detector` package's `Scorer` protocol already accommodates it; only a `detector/lstm.py` and torch in `detector`'s deps are missing.
